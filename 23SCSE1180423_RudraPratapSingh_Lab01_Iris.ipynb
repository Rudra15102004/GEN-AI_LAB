```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

```python
# Load Iris dataset
df = sns.load_dataset("iris")

print("Iris dataset loaded successfully.")
display(df)
```

```python
print("First 10 Records:")
display(df.head(10))
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Data Types of All Columns:")
print(df.dtypes)
```

```python
print("Missing Values Column-wise:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate Records:")
    display(df[df.duplicated(keep=False)])
else:
    print("No duplicate records found.")
```

## Q1 — Observation

The Iris dataset was successfully loaded into a Pandas DataFrame.

- The first 10 records were displayed.
- The dataset shape and statistical summary were obtained.
- The data types of all columns were checked.
- Missing values were identified using `isnull().sum()`.
- Duplicate records were checked using the `duplicated()` function.

QUESTION 2 — Numerical Analysis

```python
# Select numerical columns
numerical_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

statistics = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Minimum": df[numerical_columns].min(),
    "Maximum": df[numerical_columns].max()
})

display(statistics)
```

```python
print("Number of records for each species:")
print(df.groupby("species").size())
```

```python
species_averages = df.groupby("species")[[
    "sepal_length",
    "petal_length"
]].mean()

print("Average Sepal Length and Petal Length by Species:")
display(species_averages)
```

```python
highest_species = species_averages["petal_length"].idxmax()
highest_value = species_averages["petal_length"].max()

print("Species with the highest average petal length:", highest_species)
print("Highest average petal length:", round(highest_value, 2))
```

```python
flowers_with_long_petals = df[df["petal_length"] > 5.0]

print("Flowers having petal_length > 5.0:")
display(flowers_with_long_petals)
```

## Q2 — Observation

The numerical features were analyzed using mean, median, minimum, and maximum values.

The dataset was grouped according to species, and the average sepal length and petal
length were calculated for each species.

Among the three species, **virginica** has the highest average petal length.

Flowers with a petal length greater than 5.0 were also successfully filtered and displayed.

QUESTION 3 — Univariate EDA

```python
plt.figure(figsize=(8, 5))

plt.hist(df["petal_length"], bins=15, edgecolor="black")

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Petal Length Distribution Across Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.legend(title="Species")

plt.show()
```

## Task 4: Two Meaningful Observations from the Visualizations

### Observation 1: Clear Separation of Iris Setosa

The histogram of petal length shows a clear separation between *Iris setosa* and the other two species. The petal lengths of *Iris setosa* are considerably smaller, creating a distinct group. The boxplot also confirms that *Iris setosa* has much smaller petal lengths compared with *Iris versicolor* and *Iris virginica*.

### Observation 2: Positive Relationship Between Sepal Length and Petal Length

The scatter plot of sepal length versus petal length shows a positive relationship between the two features. In general, flowers with larger petal lengths also tend to have larger sepal lengths. *Iris versicolor* and *Iris virginica* show some overlap, but *Iris virginica* generally forms the group with the largest petal lengths.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student_ID": [101, 102, 103, 104, 102, 105, 106, 106],
    "Name": ["Rahul", "Anjali", "Amit", "Priya", "Anjali", "Neha", "Rohit", "Rohit"],
    "Age": [20, 21, np.nan, 22, 21, np.nan, 23, 23],
    "Gender": ["M", "F", "M", "F", "F", "F", "M", "M"],
    "Marks": [85, 90, 78, np.nan, 90, 88, 92, 92]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

Q1 — Identify Data Quality Issues

```python
print("Missing Values:")
display(df.isnull())
```

```python
print("Column-wise Missing Value Count:")
print(df.isnull().sum())
```

```python
print("Duplicate Records based on Student_ID:")
display(df[df["Student_ID"].duplicated(keep=False)].sort_values("Student_ID"))
```

### Observation

- The Age column contains 2 missing values.
- The Marks column contains 1 missing value.
- Student_ID 102 and Student_ID 106 have duplicate records.

Q2 — Handle Missing Values & Duplicates
Code cell — Calculate mean

```python
age_mean = df["Age"].mean()
marks_mean = df["Marks"].mean()

print("Mean Age:", age_mean)
print("Mean Marks:", marks_mean)
```

```python
df["Age"] = df["Age"].fillna(age_mean)
df["Marks"] = df["Marks"].fillna(marks_mean)

print("Dataset after Mean Imputation:")
display(df)
```

```python
df_cleaned = df.drop_duplicates(subset="Student_ID", keep="first")

print("Final Cleaned Dataset:")
display(df_cleaned)
```

```python
print("Missing Values after Cleaning:")
print(df_cleaned.isnull().sum())

print("\nDuplicate Student IDs after Cleaning:")
print(df_cleaned["Student_ID"].duplicated().sum())
```

### Observation

Missing values in Age and Marks were successfully replaced using mean imputation.
Duplicate records were removed based on Student_ID.

The final dataset contains no missing values and no duplicate Student_IDs.

Q3 — Univariate EDA

```python
plt.figure(figsize=(8, 5))

plt.hist(df_cleaned["Marks"], bins=5, edgecolor="black")

plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.boxplot(y=df_cleaned["Age"])

plt.title("Boxplot of Student Age")
plt.ylabel("Age")

plt.show()
```

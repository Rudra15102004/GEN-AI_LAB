**Import** **Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

**Load Titanic Dataset**

```python
# Load Titanic dataset
df = sns.load_dataset("titanic")

print("Titanic dataset loaded successfully.")
print("Dataset shape:", df.shape)

display(df)
```

**Missing Values**

```python
print("Column-wise Missing Values:")
missing_values = df.isnull().sum()

display(missing_values)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate Records:")
    display(df[df.duplicated(keep=False)])
else:
    print("No duplicate records found.")
```

**Q2 — HANDLE MISSING VALUES & DUPLICATES**

```python
# Calculate mean age
mean_age = df["age"].mean()

print("Mean Age:", mean_age)

# Fill missing Age values using mean
df["age"] = df["age"].fillna(mean_age)

print("\nMissing Age values after imputation:")
print(df["age"].isnull().sum())
```

```python
# Find mode of Embarked
embarked_mode = df["embarked"].mode()[0]

print("Mode of Embarked:", embarked_mode)

# Fill missing Embarked values
df["embarked"] = df["embarked"].fillna(embarked_mode)

print("\nMissing Embarked values after imputation:")
print(df["embarked"].isnull().sum())
```

```python
# Check duplicates before removal
print("Duplicates before removal:", df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

print("Duplicates after removal:", df.duplicated().sum())
```

```python
average_age = df["age"].mean()
average_fare = df["fare"].mean()

print("Average Age:", round(average_age, 2))
print("Average Fare:", round(average_fare, 2))
```

```python
survivors = (df["survived"] == 1).sum()
non_survivors = (df["survived"] == 0).sum()

print("Number of Survivors:", survivors)
print("Number of Non-Survivors:", non_survivors)
```

```python
survival_rate_gender = df.groupby("sex")["survived"].mean() * 100

print("Survival Rate by Gender:")
display(survival_rate_gender.round(2))
```

```python
print("Final Cleaned Dataset:")
display(df)

print("\nMissing Values After Cleaning:")
display(df.isnull().sum())

print("\nTotal Duplicate Records After Cleaning:")
print(df.duplicated().sum())
```

**## Q2 Observation**

**Missing values in the Age column were handled using mean imputation.**
**Missing values in the Embarked column were handled using mode imputation.**

**Duplicate records were removed from the dataset.**

**The average age, average fare, number of survivors, number of non-survivors,**
**and survival rate for male and female passengers were calculated successfully.**

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=20, edgecolor="black")

plt.title("Histogram of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
survival_counts = df["survived"].value_counts().sort_index()

labels = ["Non-Survivors", "Survivors"]
values = [survival_counts[0], survival_counts[1]]

plt.figure(figsize=(7, 5))

plt.bar(labels, values)

plt.title("Survivors vs Non-Survivors")
plt.xlabel("Passenger Status")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
gender_survival = df.groupby("sex")["survived"].sum()

plt.figure(figsize=(7, 5))

plt.bar(gender_survival.index, gender_survival.values)

plt.title("Survival Counts by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Survivors")

plt.show()
```
